```python
marks = [45, 67, 82, 91, 39, 75, 58]

result = []

for mark in marks:
    if mark > 60:
        result.append(mark)

print(result)
```

```python
import numpy as np

marks = np.array([45, 67, 82, 91, 39, 75, 58])

print(marks > 60)
```

```python
print(marks[marks > 60])
```

```python
marks = np.array([45,67,82,91,39,75,58])
print(marks[(marks > 60) & (marks < 90)])
```
