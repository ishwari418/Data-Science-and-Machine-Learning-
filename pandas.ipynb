```python
maze = {
    'Start': ['A', 'F'],
    'A': ['Start', 'F', 'E'],
    'B': ['E','Goal'],
    'C': ['D','E'],
    'D': ['C', 'Goal'],
    'E': ['B','A', 'C'],
    'F': ['A'],
    'Goal': ['B', 'D']
}

print(type(maze))
print("Maze Environment Successfully Created")
print(maze)
```

```python
def solve_bfs(graph, start, goal):
    queue = [[start]]      # Queue stores complete paths
    visited = set()

    while queue:
        path = queue.pop(0)
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbor in graph.get(node, []):
                new_path = path.copy()   
                new_path.append(neighbor)
                queue.append(new_path)

    return "No path found"


bfs_path = solve_bfs(maze, 'Start', 'Goal')
print("BFS Path from Start to Goal:", bfs_path)
```

```python
def solve_dfs(graph,start,goal):
    stack=[[start]]
    visited=set()
    while stack:
        path=stack.pop()
        node=path[-1]
        if node==goal:
            return path
        if node not in visited:
            visited.add(node)
            for neighbor in graph.get(node,[]):
                new_path=path.copy()
                new_path.append(neighbor)
                stack.append(new_path)
    return "No path found"
dfs_path=solve_dfs(maze,'Start','Goal')
print("DFS Path from Start to Goal:",dfs_path)
```
