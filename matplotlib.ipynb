```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
product_a = [100, 120, 150, 170]
product_b = [90, 110, 130, 160]

plt.plot(months, product_a, label="Product A")
plt.plot(months, product_b, label="Product B")

plt.title("Monthly Sales Comparison")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid()

plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,8))

# 1. Line Chart
plt.subplot(2,2,1)
plt.plot([1,2,3,4], [10,20,15,30])
plt.title("Accuracy Trend")

# 2. Bar Chart
plt.subplot(2,2,2)
plt.bar(["A","B","C"], [25,40,30])
plt.title("Disease Count")

# 3. Pie Chart
plt.subplot(2,2,3)
plt.pie([40,60], labels=["Healthy","Disease"], autopct="%1.1f%%")
plt.title("Patient Distribution")

# 4. Histogram
plt.subplot(2,2,4)
plt.hist([20,25,22,24,30,35,28,32])
plt.title("Age Distribution")

plt.tight_layout()
plt.show()
```
