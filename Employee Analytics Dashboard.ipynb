```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    "Employee_ID": [101,102,103,104,105,106,107,108,109,110],
    "Department": ["IT","HR","IT","Sales","HR","IT","Sales","IT","Marketing","HR"],
    "Gender": ["Male","Female","Male","Female","Female","Male","Male","Female","Male","Female"],
    "Age": [25,30,28,35,27,32,29,26,31,33],
    "Salary": [45000,52000,48000,60000,50000,70000,55000,47000,62000,58000],
    "Experience": [2,5,3,8,4,9,6,2,7,5]
}

df = pd.DataFrame(data)
```

```python
sns.countplot(data=df, x="Department")
plt.title("Employees in Each Department")
plt.show()
```

```python
sns.barplot(data=df, x="Department", y="Salary")
plt.title("Average Salary by Department")
plt.show()
```

```python
sns.histplot(data=df, x="Salary", bins=5)
plt.title("Salary Distribution")
plt.show()
```

```python
sns.scatterplot(data=df, x="Experience", y="Salary")
plt.title("Experience vs Salary")
plt.show()
```

```python
sns.boxplot(data=df, x="Department", y="Salary")
plt.title("Salary Comparison by Department")
plt.show()
```

```python
corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```
