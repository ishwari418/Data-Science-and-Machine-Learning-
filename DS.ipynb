Slicing

```python
numbers=[23,65,44,55,33,876,22]
print(numbers[:3])
```

```python
numbers=[23,65,44,55,33,876,22]
print (numbers[::-1])
print(numbers[1:4])
print (numbers[::-2])
print (numbers[2:])
print(numbers[-3::])
print(len(numbers))
```
